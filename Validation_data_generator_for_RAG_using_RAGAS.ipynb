```python
#Load documents
from langchain_community.document_loaders import DirectoryLoader
path = "Sample_Docs_Markdown/"
loader = DirectoryLoader(path, glob="**/*.md")
docs = loader.load()
print(len(docs), os.listdir(path), len(os.listdir(path)))
```

```python
%%time
from langchain_huggingface.embeddings import HuggingFaceEmbeddings
model_name = "sentence-transformers/all-MiniLM-L6-v2"
embeddings = HuggingFaceEmbeddings(
    model_name=model_name,
    model_kwargs={'device': 'cuda'},  #Use 'cuda' if you have a GPU
    encode_kwargs={'normalize_embeddings': False} #Set to True for cosine similarity
)

#sample test
text = "This is a test document to generate an embedding for."
query_embedding = embeddings.embed_query(text)
print(f"Query embedding (first 5 values): {query_embedding[:5]}...")
print(f"Size of embedding : {len(query_embedding)}")
```

```python
from ragas.embeddings import LangchainEmbeddingsWrapper # to use custom embedding model : Default: OpenAIEmbeddings
from ragas.llms import LangchainLLMWrapper # to use custom LLM model : Default: OpenAIModel

generator_llm = LangchainLLMWrapper(llm)  #use better model & better embeddding for better results
custom_embedding_model= LangchainLLMWrapper(embeddings)
```

```python
from ragas.testset import TestsetGenerator

generator = TestsetGenerator(llm=generator_llm,
                             embedding_model=custom_embedding_model)


dataset = generator.generate_with_langchain_docs(docs, testset_size=10)
```
