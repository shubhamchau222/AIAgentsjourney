```python
import threading
import time
```

```python
##Indicate some task being done
def func(second):
    print(f"Sleeping for {second} seconds..")
    time.sleep(second)

func(4)
func(7)
func(2)
```

```python
## Start all above funct in parallel
## Code using threads

t1= threading.Thread(target=func, args=(4,))
t2= threading.Thread(target=func, args=(7,))
t3= threading.Thread(target=func, args=(2,))

t1.start()
t2.start()
t3.start()
#but it should take 7 seconds at list
#because when calling t1, t2, t3 it;ll start executing in background
```

```python
t1= threading.Thread(target=func, args=(4,))
t2= threading.Thread(target=func, args=(7,))
t3= threading.Thread(target=func, args=(2,))

t1.start()
t2.start()
t3.start()

t1.join()
t2.join()
t3.join()

## Now it'll take 7conds max
```

```python
def main():
    t1= threading.Thread(target=func, args=(4,))
    t2= threading.Thread(target=func, args=(7,))
    t3= threading.Thread(target=func, args=(2,))

    t1.start()
    t2.start()
    t3.start()

    t1.join()
    t2.join()
    t3.join()
```

```python
from concurrent.futures import ThreadPoolExecutor

with ThreadPoolExecutor() as executor:
    future1= executor.submit(func, 2)
    future2= executor.submit(func, 4)
    future2= executor.submit(func, 7)

    print(future1.result())
```

```python
#Threadppull Executor with mapping functions
def func(second):
    print(f"Sleeping for {second} seconds..")
    time.sleep(second)
    return second


with ThreadPoolExecutor() as executor:
    l= [4,7,8,12]
    results= executor.map(func, l)
    for result in results:
        print(result)
```

* It enhances speed by utilizing wait times for other tasks, improving efficiency.
* It enables handling numerous tasks or users concurrently without performance degradation.
* It maintains responsiveness by preventing program freezes during waiting periods.

* The async Keyword

* The **async** keyword helps you create functions in Python that can do several things at once, without waiting in line. 
* Think of it like a chef who starts cooking multiple dishes simultaneously, not just one at a time

```python
async def fetch_data():
    response= await print("hello")
    return response
```

**Important points about async:**

* It marks a function as ready to handle several tasks together.
* These functions can take a break with `await` when they're waiting for something.
* When you call an async function, it prepares but doesn't start right away.
* To actually run it, you use `asyncio.run()`.

```python
asyncio.run(fetch_data())
```

* Too many await can slow things down, so balance is key
* To wait on several tasks at once, asyncio.gather() is your friend.


```python
import asyncio

async def func1():
    time.sleep(2)
    print('func1 complete')

async def func2():
    time.sleep(4)
    print('func2 complete')

await asyncio.gather(func1(), func2())
```

* What asyncio.gather() does:

* Runs many async functions at the same time.
* Puts all the results in a list for you.
* If something goes wrong, it lets you know.
* You don't have to keep track of everything yourself.
